# Testing usb_device.py program

```python
from usb_device import get_product_name, safe_get_string, list_usb_devices, get_known_product_name, list_webcam_devices
```

```python
webcam_devices = list_webcam_devices()

print("Webcam Devices:")
webcam = []
index = 0
for wd in webcam_devices:
    print(f"webcam({index}):")
    webcam.append(wd)
    index += 1
print("Collected Webcam Devices:")
for i, wd in enumerate(webcam):
    print(f"webcam({i}): {wd}")
```

```python
for wd in webcam_devices:
    vid = wd["vendor_id"]
    pid = wd["product_id"]
    print(vid, pid)
    product_name = get_known_product_name(vid, pid)
    print(f"Product name for VID:PID: {product_name}")
```
